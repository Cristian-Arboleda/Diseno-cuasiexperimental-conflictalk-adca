H₀: Los datos provienen de una distribución normal

H₁: Los datos no provienen de una distribución normal

```python
test = 'pretest'

# Variables que se evaluarán
variables = ['Auto-asertividad', 'Hetero-asertividad']

# Filtrar los datos
datos_filtrados = estilos_aplicados[
    estilos_aplicados['Test'].str.lower() == test
]

resultados = []

for variable in variables:

    for grupo in ['Experimental', 'Control']:

        datos = datos_filtrados.loc[
            datos_filtrados['Grupo'] == grupo,
            variable
        ].dropna()

        n = len(datos)
        media = datos.mean()
        desviacion = datos.std(ddof=1)

        # Prueba de normalidad
        W, p_value = shapiro(datos)

        alpha = 0.05

        if p_value <= alpha:
            decision = 'Se rechaza H₀'
            distribucion = 'No normal'
        else:
            decision = 'No se rechaza H₀'
            distribucion = 'Normal'

        # Intervalo de confianza del 95% para la media
        error = t.ppf(1 - alpha / 2, df=n - 1) * (
            desviacion / np.sqrt(n)
        )

        resultados.append({
            'Variable': variable,
            'Test': test.capitalize(),
            'Grupo': grupo,
            'N': n,
            'Media': round(media, 3),
            'Desviación Estándar': round(desviacion, 3),
            'Prueba': 'Shapiro-Wilk',
            'IC 95% inferior': round(media - error, 3),
            'IC 95% superior': round(media + error, 3),
            'α': alpha,
            'p-value': round(p_value, 4),
            'W': round(W, 4),
            'Resultado hipótesis': decision,
            'Distribución': distribucion
        })

tabla_asertividad = pd.DataFrame(resultados)

tabla_asertividad
```

Prueba	H₀	H₁

t de Student / Welch	Las medias de ambos grupos son iguales	Las medias son diferentes

Mann-Whitney	Las distribuciones de ambos grupos son iguales	Las distribuciones son diferentes

Levene	Las varianzas de ambos grupos son iguales	Las varianzas son diferentes


```python
# Parámetros
alpha = 0.05

variables = ['Auto-asertividad', 'Hetero-asertividad']
grupos = ['Experimental', 'Control']

# Seleccionar test
test = 'pretest'

# Filtrar datos
datos = estilos_aplicados[
    estilos_aplicados['Test'].astype(str).str.strip().str.lower() == test
].copy()

# Normalizar Grupo
datos['Grupo'] = (
    datos['Grupo']
    .astype(str)
    .str.strip()
    .str.lower()
)

resultados = []

# Recorrer variables
for variable in variables:

    # Obtener datos de cada grupo
    experimental = datos.loc[
        datos['Grupo'] == 'experimental',
        variable
    ].dropna()

    control = datos.loc[
        datos['Grupo'] == 'control',
        variable
    ].dropna()

    # Pruebas de normalidad
    W_exp, p_exp = shapiro(experimental)
    W_control, p_control = shapiro(control)

    # Clasificar distribuciones
    if p_exp > alpha and p_control > alpha:
        distribuciones = 'Normal - Normal'
        enfoque = 'Paramétrica'

        # Prueba de homogeneidad de varianzas
        estadistico_levene, p_levene = levene(
            experimental,
            control
        )

        if p_levene > alpha:
            homogeneidad = 'Homogéneas'

            # t de Student
            estadistico, pvalue = ttest_ind(
                experimental,
                control,
                equal_var=True
            )

            prueba = 't de Student'
            gl = len(experimental) + len(control) - 2

        else:
            homogeneidad = 'Heterogéneas'

            # t de Welch
            estadistico, pvalue = ttest_ind(
                experimental,
                control,
                equal_var=False
            )

            prueba = 't de Welch'

            # Grados de libertad Welch-Satterthwaite
            s1 = experimental.var(ddof=1)
            s2 = control.var(ddof=1)
            n1 = len(experimental)
            n2 = len(control)

            gl = (
                (s1/n1 + s2/n2)**2
                /
                (
                    (s1/n1)**2/(n1-1)
                    +
                    (s2/n2)**2/(n2-1)
                )
            )

        tipo_grupos = 'Independientes'
        prueba_homogeneidad = 'Levene'

    else:
        distribuciones = 'No normal - Normal' if (
            (p_exp <= alpha) != (p_control <= alpha)
        ) else 'No normal - No normal'

        enfoque = 'No paramétrica'

        # Mann-Whitney
        estadistico, pvalue = mannwhitneyu(
            experimental,
            control,
            alternative='two-sided'
        )

        prueba = 'U de Mann-Whitney'
        tipo_grupos = 'Independientes'
        prueba_homogeneidad = 'No aplica'
        estadistico_levene = np.nan
        p_levene = np.nan
        homogeneidad = 'No aplica'
        gl = np.nan

    # Decisión de hipótesis
    if pvalue <= alpha:
        decision = 'Se rechaza H₀'
    else:
        decision = 'No se rechaza H₀'

    # Agregar resultados
    resultados.append({
        'Variable': variable,
        'Grupos comparados': 'Experimental - Control',
        'Test': test.capitalize(),
        'Distribuciones': distribuciones,
        'Enfoque estadístico': enfoque,
        'Tipo de grupos': tipo_grupos,
        'Prueba de homogeneidad': prueba_homogeneidad,
        'Estadístico Levene': estadistico_levene if enfoque == 'Paramétrica' else np.nan,
        'p-valor Levene': p_levene if enfoque == 'Paramétrica' else np.nan,
        'Homogeneidad de varianzas': homogeneidad,
        'Prueba aplicada': prueba,
        'Estadístico': estadistico,
        'gl': gl,
        'α': alpha,
        'pvalue': pvalue,
        'Decisión sobre H₀': decision
    })

# Crear tabla
tabla_comparaciones = pd.DataFrame(resultados)

# Redondear resultados numéricos
columnas_numericas = [
    'Estadístico Levene',
    'p-valor Levene',
    'Estadístico',
    'gl',
    'α',
    'pvalue'
]

tabla_comparaciones[columnas_numericas] = (
    tabla_comparaciones[columnas_numericas].round(4)
)

tabla_comparaciones
```

# 2. Efecto del tratamiento <br>
**Comparación:** Experimental (Pretest | Postest) <br>
**Intención:**  Saber si la intervención tuvo un efecto estadisticamente significativo en la poblacion objetivo

H₀: Los datos provienen de una distribución normal

H₁: Los datos no provienen de una distribución normal

```python

# Parámetros
alpha = 0.05

# Seleccionar grupo
grupo = 'experimental'

# Variables a evaluar
variables = ['Auto-asertividad', 'Hetero-asertividad']

# Normalizar categorías
datos = estilos_aplicados.copy()

for columna in ['Grupo', 'Test']:
    datos[columna] = (
        datos[columna]
        .astype(str)
        .str.strip()
        .str.lower()
    )

# Filtrar grupo seleccionado
datos_grupo = datos[
    datos['Grupo'] == grupo.lower()
].copy()

# Separar Pretest y Postest
pretest = datos_grupo[
    datos_grupo['Test'] == 'pretest'
]

postest = datos_grupo[
    datos_grupo['Test'] == 'postest'
]

# Emparejar mediante ID
datos_pareados = pretest.merge(
    postest,
    on='ID',
    suffixes=('_Pretest', '_Postest')
)

resultados = []

# Recorrer variables
for variable in variables:

    # Obtener diferencias individuales
    cambios = (
        datos_pareados[f'{variable}_Postest']
        -
        datos_pareados[f'{variable}_Pretest']
    ).dropna()

    n = len(cambios)

    media = cambios.mean()
    desviacion = cambios.std(ddof=1)

    # Prueba de normalidad sobre diferencias
    resultado_shapiro = shapiro(cambios)

    W = resultado_shapiro.statistic
    pvalue = resultado_shapiro.pvalue

    # Decisión de normalidad
    if pvalue > alpha:
        decision = 'No se rechaza H₀'
        distribucion = 'Normal'
    else:
        decision = 'Se rechaza H₀'
        distribucion = 'No normal'

    # Intervalo de confianza del 95% para el cambio medio
    error = t.ppf(
        1 - alpha / 2,
        df=n - 1
    ) * (desviacion / np.sqrt(n))

    ic_inferior = media - error
    ic_superior = media + error

    # Guardar resultados
    resultados.append({
        'Variable': variable,
        'Momentos Comparados': 'Pretest - Postest',
        'Grupo': grupo.capitalize(),
        'N': n,
        'Media': round(media, 4),
        'Desviación Estándar': round(desviacion, 4),
        'Prueba': 'Shapiro-Wilk',
        'IC 95% inferior': round(ic_inferior, 4),
        'IC 95% superior': round(ic_superior, 4),
        'α': alpha,
        'p-value': round(pvalue, 4),
        'W': round(W, 4),
        'Resultado hipótesis': decision,
        'Distribución': distribucion
    })

# Crear tabla
tabla_normalidad_cambios = pd.DataFrame(resultados)

tabla_normalidad_cambios
```

Hipótesis de las pruebas aplicadas
Para la prueba t de Student pareada:

- H₀: la media de las diferencias entre Pretest y Postest es igual a cero.

- H₁: la media de las diferencias es distinta de cero.

Para Wilcoxon:

- H₀: no existe un cambio sistemático entre Pretest y Postest.

- H₁: existe un cambio sistemático entre ambos momentos.

En ambas pruebas, con alpha=0.05:

- p>0.05: no se rechaza H₀.

- p<=0.05: se rechaza H₀.

```python
alpha = 0.05

# Seleccionar grupo
grupo = 'experimental'

# Variables de asertividad
variables = ['Auto-asertividad', 'Hetero-asertividad']

# Normalizar los datos
datos = estilos_aplicados.copy()

datos['Grupo'] = (
    datos['Grupo']
    .astype(str)
    .str.strip()
    .str.lower()
)

datos['Test'] = (
    datos['Test']
    .astype(str)
    .str.strip()
    .str.lower()
)

# Convertir puntajes a valores numéricos
for variable in variables:
    datos[variable] = pd.to_numeric(
        datos[variable],
        errors='coerce'
    )

# Filtrar grupo seleccionado
datos_grupo = datos[
    datos['Grupo'] == grupo.lower()
].copy()

resultados_pruebas = []

# Recorrer variables de asertividad
for variable in variables:

    # Emparejar participantes por ID
    datos_pareados = datos_grupo.pivot(
        index='ID',
        columns='Test',
        values=variable
    )

    datos_pareados = datos_pareados.reindex(
        columns=['pretest', 'postest']
    ).dropna()

    n = len(datos_pareados)

    # Calcular diferencias Postest - Pretest
    diferencias = (
        datos_pareados['postest'] -
        datos_pareados['pretest']
    )

    # Valores iniciales
    distribucion = 'No calculable'
    W = np.nan
    p_normalidad = np.nan

    prueba_aplicada = 'No calculable'
    enfoque = 'No calculable'
    estadistico = np.nan
    gl = 'No aplica'
    pvalue = np.nan
    decision = 'No calculable'

    # Evaluar normalidad
    if 3 <= n <= 5000:

        resultado_normalidad = shapiro(diferencias)

        W = resultado_normalidad.statistic
        p_normalidad = resultado_normalidad.pvalue

        if p_normalidad > alpha:
            distribucion = 'Normal'
        else:
            distribucion = 'No normal'

    # Aplicar prueba estadística
    if distribucion == 'Normal':

        prueba = ttest_rel(
            datos_pareados['postest'],
            datos_pareados['pretest']
        )

        prueba_aplicada = 't de Student pareada'
        enfoque = 'Paramétrico'
        estadistico = prueba.statistic
        gl = n - 1
        pvalue = prueba.pvalue

    elif distribucion == 'No normal':

        try:
            prueba = wilcoxon(
                datos_pareados['postest'],
                datos_pareados['pretest'],
                alternative='two-sided',
                method='auto'
            )

            prueba_aplicada = 'Wilcoxon'
            enfoque = 'No paramétrico'
            estadistico = prueba.statistic
            pvalue = prueba.pvalue

        except ValueError:
            prueba_aplicada = 'Wilcoxon'
            enfoque = 'No paramétrico'

    # Decisión estadística
    if pd.notna(pvalue):

        if pvalue > alpha:
            decision = 'No se rechaza H₀'
        else:
            decision = 'Se rechaza H₀'

    # Guardar resultados
    resultados_pruebas.append({
        'Variable': variable,
        'Momentos Comparados': 'Pretest - Postest',
        'Grupo': grupo.capitalize(),
        'n': n,
        'Distribución': distribucion,
        'W Shapiro': W,
        'p Shapiro': p_normalidad,
        'Enfoque estadístico': enfoque,
        'Tipo de grupos': 'Relacionados',
        'Prueba aplicada': prueba_aplicada,
        'Estadístico': estadistico,
        'gl': gl,
        'α': alpha,
        'p-value': pvalue,
        'Decisión sobre H₀': decision
    })

# Crear DataFrame
tabla_pruebas = pd.DataFrame(resultados_pruebas)

# Ordenar variables
orden_variables = pd.CategoricalDtype(
    categories=[
        'Auto-asertividad',
        'Hetero-asertividad'
    ],
    ordered=True
)

tabla_pruebas['Variable'] = (
    tabla_pruebas['Variable'].astype(orden_variables)
)

tabla_pruebas = (
    tabla_pruebas
    .sort_values('Variable')
    .reset_index(drop=True)
)

# Mostrar tabla
tabla_pruebas.style.format({
    'W Shapiro': '{:.4f}',
    'p Shapiro': '{:.4f}',
    'Estadístico': '{:.4f}',
    'α': '{:.2f}',
    'p-value': '{:.4f}'
})
```

# 3. Estabilidad sin tratamiento<br>
**Comparación:** Control (Pretest | Postest) <br>
**Intención:**  Evaluar la estabilida de las diferencias del grupo sin tratamiento

Prueba para la distribución de datos

H₀: Los datos provienen de una distribución normal

H₁: Los datos no provienen de una distribución normal

```python

# Parámetros
alpha = 0.05

# Seleccionar grupo
grupo = 'control'

# Variables a evaluar
variables = ['Auto-asertividad', 'Hetero-asertividad']

# Normalizar categorías
datos = estilos_aplicados.copy()

for columna in ['Grupo', 'Test']:
    datos[columna] = (
        datos[columna]
        .astype(str)
        .str.strip()
        .str.lower()
    )

# Filtrar grupo seleccionado
datos_grupo = datos[
    datos['Grupo'] == grupo.lower()
].copy()

# Separar Pretest y Postest
pretest = datos_grupo[
    datos_grupo['Test'] == 'pretest'
]

postest = datos_grupo[
    datos_grupo['Test'] == 'postest'
]

# Emparejar mediante ID
datos_pareados = pretest.merge(
    postest,
    on='ID',
    suffixes=('_Pretest', '_Postest')
)

resultados = []

# Recorrer variables
for variable in variables:

    # Obtener diferencias individuales
    cambios = (
        datos_pareados[f'{variable}_Postest']
        -
        datos_pareados[f'{variable}_Pretest']
    ).dropna()

    n = len(cambios)

    media = cambios.mean()
    desviacion = cambios.std(ddof=1)

    # Prueba de normalidad sobre diferencias
    resultado_shapiro = shapiro(cambios)

    W = resultado_shapiro.statistic
    pvalue = resultado_shapiro.pvalue

    # Decisión de normalidad
    if pvalue > alpha:
        decision = 'No se rechaza H₀'
        distribucion = 'Normal'
    else:
        decision = 'Se rechaza H₀'
        distribucion = 'No normal'

    # Intervalo de confianza del 95% para el cambio medio
    error = t.ppf(
        1 - alpha / 2,
        df=n - 1
    ) * (desviacion / np.sqrt(n))

    ic_inferior = media - error
    ic_superior = media + error

    # Guardar resultados
    resultados.append({
        'Variable': variable,
        'Momentos Comparados': 'Pretest - Postest',
        'Grupo': grupo.capitalize(),
        'N': n,
        'Media': round(media, 4),
        'Desviación Estándar': round(desviacion, 4),
        'Prueba': 'Shapiro-Wilk',
        'IC 95% inferior': round(ic_inferior, 4),
        'IC 95% superior': round(ic_superior, 4),
        'α': alpha,
        'p-value': round(pvalue, 4),
        'W': round(W, 4),
        'Resultado hipótesis': decision,
        'Distribución': distribucion
    })

# Crear tabla
tabla_normalidad_cambios = pd.DataFrame(resultados)

tabla_normalidad_cambios
```

```python
alpha = 0.05

# Seleccionar grupo
grupo = 'control'

# Variables de asertividad
variables = ['Auto-asertividad', 'Hetero-asertividad']

# Normalizar los datos
datos = estilos_aplicados.copy()

datos['Grupo'] = (
    datos['Grupo']
    .astype(str)
    .str.strip()
    .str.lower()
)

datos['Test'] = (
    datos['Test']
    .astype(str)
    .str.strip()
    .str.lower()
)

# Convertir puntajes a valores numéricos
for variable in variables:
    datos[variable] = pd.to_numeric(
        datos[variable],
        errors='coerce'
    )

# Filtrar grupo seleccionado
datos_grupo = datos[
    datos['Grupo'] == grupo.lower()
].copy()

resultados_pruebas = []

# Recorrer variables de asertividad
for variable in variables:

    # Emparejar participantes por ID
    datos_pareados = datos_grupo.pivot(
        index='ID',
        columns='Test',
        values=variable
    )

    datos_pareados = datos_pareados.reindex(
        columns=['pretest', 'postest']
    ).dropna()

    n = len(datos_pareados)

    # Calcular diferencias Postest - Pretest
    diferencias = (
        datos_pareados['postest'] -
        datos_pareados['pretest']
    )

    # Valores iniciales
    distribucion = 'No calculable'
    W = np.nan
    p_normalidad = np.nan

    prueba_aplicada = 'No calculable'
    enfoque = 'No calculable'
    estadistico = np.nan
    gl = 'No aplica'
    pvalue = np.nan
    decision = 'No calculable'

    # Evaluar normalidad
    if 3 <= n <= 5000:

        resultado_normalidad = shapiro(diferencias)

        W = resultado_normalidad.statistic
        p_normalidad = resultado_normalidad.pvalue

        if p_normalidad > alpha:
            distribucion = 'Normal'
        else:
            distribucion = 'No normal'

    # Aplicar prueba estadística
    if distribucion == 'Normal':

        prueba = ttest_rel(
            datos_pareados['postest'],
            datos_pareados['pretest']
        )

        prueba_aplicada = 't de Student pareada'
        enfoque = 'Paramétrico'
        estadistico = prueba.statistic
        gl = n - 1
        pvalue = prueba.pvalue

    elif distribucion == 'No normal':

        try:
            prueba = wilcoxon(
                datos_pareados['postest'],
                datos_pareados['pretest'],
                alternative='two-sided',
                method='auto'
            )

            prueba_aplicada = 'Wilcoxon'
            enfoque = 'No paramétrico'
            estadistico = prueba.statistic
            pvalue = prueba.pvalue

        except ValueError:
            prueba_aplicada = 'Wilcoxon'
            enfoque = 'No paramétrico'

    # Decisión estadística
    if pd.notna(pvalue):

        if pvalue > alpha:
            decision = 'No se rechaza H₀'
        else:
            decision = 'Se rechaza H₀'

    # Guardar resultados
    resultados_pruebas.append({
        'Variable': variable,
        'Momentos Comparados': 'Pretest - Postest',
        'Grupo': grupo.capitalize(),
        'n': n,
        'Distribución': distribucion,
        'W Shapiro': W,
        'p Shapiro': p_normalidad,
        'Enfoque estadístico': enfoque,
        'Tipo de grupos': 'Relacionados',
        'Prueba aplicada': prueba_aplicada,
        'Estadístico': estadistico,
        'gl': gl,
        'α': alpha,
        'p-value': pvalue,
        'Decisión sobre H₀': decision
    })

# Crear DataFrame
tabla_pruebas = pd.DataFrame(resultados_pruebas)

# Ordenar variables
orden_variables = pd.CategoricalDtype(
    categories=[
        'Auto-asertividad',
        'Hetero-asertividad'
    ],
    ordered=True
)

tabla_pruebas['Variable'] = (
    tabla_pruebas['Variable'].astype(orden_variables)
)

tabla_pruebas = (
    tabla_pruebas
    .sort_values('Variable')
    .reset_index(drop=True)
)

# Mostrar tabla
tabla_pruebas.style.format({
    'W Shapiro': '{:.4f}',
    'p Shapiro': '{:.4f}',
    'Estadístico': '{:.4f}',
    'α': '{:.2f}',
    'p-value': '{:.4f}'
})
```

```python
# Parámetros
alpha = 0.05

variables = ['Auto-asertividad', 'Hetero-asertividad']
grupos = ['Experimental', 'Control']

# Seleccionar test
test = 'postest'

# Filtrar datos
datos = estilos_aplicados[
    estilos_aplicados['Test'].astype(str).str.strip().str.lower() == test
].copy()

# Normalizar Grupo
datos['Grupo'] = (
    datos['Grupo']
    .astype(str)
    .str.strip()
    .str.lower()
)

resultados = []

# Recorrer variables
for variable in variables:

    # Obtener datos de cada grupo
    experimental = datos.loc[
        datos['Grupo'] == 'experimental',
        variable
    ].dropna()

    control = datos.loc[
        datos['Grupo'] == 'control',
        variable
    ].dropna()

    # Pruebas de normalidad
    W_exp, p_exp = shapiro(experimental)
    W_control, p_control = shapiro(control)

    # Clasificar distribuciones
    if p_exp > alpha and p_control > alpha:
        distribuciones = 'Normal - Normal'
        enfoque = 'Paramétrica'

        # Prueba de homogeneidad de varianzas
        estadistico_levene, p_levene = levene(
            experimental,
            control
        )

        if p_levene > alpha:
            homogeneidad = 'Homogéneas'

            # t de Student
            estadistico, pvalue = ttest_ind(
                experimental,
                control,
                equal_var=True
            )

            prueba = 't de Student'
            gl = len(experimental) + len(control) - 2

        else:
            homogeneidad = 'Heterogéneas'

            # t de Welch
            estadistico, pvalue = ttest_ind(
                experimental,
                control,
                equal_var=False
            )

            prueba = 't de Welch'

            # Grados de libertad Welch-Satterthwaite
            s1 = experimental.var(ddof=1)
            s2 = control.var(ddof=1)
            n1 = len(experimental)
            n2 = len(control)

            gl = (
                (s1/n1 + s2/n2)**2
                /
                (
                    (s1/n1)**2/(n1-1)
                    +
                    (s2/n2)**2/(n2-1)
                )
            )

        tipo_grupos = 'Independientes'
        prueba_homogeneidad = 'Levene'

    else:
        distribuciones = 'No normal - Normal' if (
            (p_exp <= alpha) != (p_control <= alpha)
        ) else 'No normal - No normal'

        enfoque = 'No paramétrica'

        # Mann-Whitney
        estadistico, pvalue = mannwhitneyu(
            experimental,
            control,
            alternative='two-sided'
        )

        prueba = 'U de Mann-Whitney'
        tipo_grupos = 'Independientes'
        prueba_homogeneidad = 'No aplica'
        estadistico_levene = np.nan
        p_levene = np.nan
        homogeneidad = 'No aplica'
        gl = np.nan

    # Decisión de hipótesis
    if pvalue <= alpha:
        decision = 'Se rechaza H₀'
    else:
        decision = 'No se rechaza H₀'

    # Agregar resultados
    resultados.append({
        'Variable': variable,
        'Grupos comparados': 'Experimental - Control',
        'Test': test.capitalize(),
        'Distribuciones': distribuciones,
        'Enfoque estadístico': enfoque,
        'Tipo de grupos': tipo_grupos,
        'Prueba de homogeneidad': prueba_homogeneidad,
        'Estadístico Levene': estadistico_levene if enfoque == 'Paramétrica' else np.nan,
        'p-valor Levene': p_levene if enfoque == 'Paramétrica' else np.nan,
        'Homogeneidad de varianzas': homogeneidad,
        'Prueba aplicada': prueba,
        'Estadístico': estadistico,
        'gl': gl,
        'α': alpha,
        'pvalue': pvalue,
        'Decisión sobre H₀': decision
    })

# Crear tabla
tabla_comparaciones = pd.DataFrame(resultados)

# Redondear resultados numéricos
columnas_numericas = [
    'Estadístico Levene',
    'p-valor Levene',
    'Estadístico',
    'gl',
    'α',
    'pvalue'
]

tabla_comparaciones[columnas_numericas] = (
    tabla_comparaciones[columnas_numericas].round(4)
)

tabla_comparaciones
```

# 5.Efecto diferencial de la intervención <br>
**Comparación:** Pretest (Control | Experimental) - Postest (Control | Experimental) <br>
**Intención:** Evaluar los cambios de ambos grupos

```python
import pandas as pd

variables = [
    'Auto-asertividad',
    'Hetero-asertividad'
]

# Copiar los datos
datos = estilos_aplicados.copy()

# Normalizar etiquetas
datos['Grupo'] = (
    datos['Grupo']
    .astype(str)
    .str.strip()
    .str.lower()
)

datos['Test'] = (
    datos['Test']
    .astype(str)
    .str.strip()
    .str.lower()
)

# Convertir puntajes a valores numéricos
for variable in variables:
    datos[variable] = pd.to_numeric(
        datos[variable],
        errors='coerce'
    )

# Guardar cambios individuales
lista_cambios = []

for variable in variables:

    # Organizar Pretest y Postest por participante
    tabla = datos.pivot(
        index=['Grupo', 'ID'],
        columns='Test',
        values=variable
    ).reset_index()

    # Calcular cambio individual
    tabla['Cambio'] = (
        tabla['postest'] - tabla['pretest']
    )

    # Identificar variable
    tabla['Variable'] = variable

    lista_cambios.append(tabla)

# Unir los cambios de ambas variables
tabla_cambios = pd.concat(
    lista_cambios,
    ignore_index=True
)

# Ordenar columnas
tabla_cambios = tabla_cambios[
    [
        'Grupo',
        'ID',
        'Variable',
        'pretest',
        'postest',
        'Cambio'
    ]
]

# Ordenar resultados
tabla_cambios = tabla_cambios.sort_values(
    ['Grupo', 'ID', 'Variable']
).reset_index(drop=True)

# Mostrar tabla
tabla_cambios
```

t de Welch: comparación de medias
Hipótesis nula (H₀):

No existe diferencia entre los cambios promedio de ambos grupos.

Hipótesis alternativa (H₁):

Existe una diferencia entre los cambios promedio de ambos grupos.

Hipótesis nula (H₀):

U de Mann-Whitney: comparación de distribuciones

Las distribuciones de los cambios del grupo Control y Experimental son iguales.

Hipótesis alternativa (H₁):

Las distribuciones de los cambios de ambos grupos son diferentes.

```python
from scipy.stats import shapiro, ttest_ind, mannwhitneyu

alpha = 0.05

variables = ['Auto-asertividad', 'Hetero-asertividad']

resultados = []

for variable in variables:

    # Extraer los cambios de cada grupo
    control = tabla_cambios.loc[
        (tabla_cambios['Grupo'] == 'control') &
        (tabla_cambios['Variable'] == variable),
        'Cambio'
    ].dropna()

    experimental = tabla_cambios.loc[
        (tabla_cambios['Grupo'] == 'experimental') &
        (tabla_cambios['Variable'] == variable),
        'Cambio'
    ].dropna()

    # Comprobar normalidad de los cambios
    W_control, p_control = shapiro(control)
    W_exp, p_exp = shapiro(experimental)

    # Seleccionar la prueba estadística
    if p_control > alpha and p_exp > alpha:

        estadistico, p_value = ttest_ind(
            experimental,
            control,
            equal_var=False
        )

        prueba = 't de Welch'
        enfoque = 'Paramétrico'

    else:

        estadistico, p_value = mannwhitneyu(
            experimental,
            control,
            alternative='two-sided'
        )

        prueba = 'U de Mann-Whitney'
        enfoque = 'No paramétrico'

    # Diferencia entre los cambios promedio
    diferencia_cambios = (
        experimental.mean() - control.mean()
    )

    # Decisión sobre la hipótesis
    decision = (
        'Se rechaza H₀'
        if p_value <= alpha
        else 'No se rechaza H₀'
    )

    resultados.append({
        'Variable': variable,
        'N Control': len(control),
        'N Experimental': len(experimental),
        'Shapiro Control': p_control,
        'Shapiro Experimental': p_exp,
        'Cambio promedio Control': control.mean(),
        'Cambio promedio Experimental': experimental.mean(),
        'Diferencia de cambios': diferencia_cambios,
        'Enfoque estadístico': enfoque,
        'Prueba aplicada': prueba,
        'Estadístico': estadistico,
        'α': alpha,
        'p-value': p_value,
        'Decisión sobre H₀': decision
    })

tabla_efecto = pd.DataFrame(resultados)

tabla_efecto
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

variables = ['Auto-asertividad', 'Hetero-asertividad']
grupos = ['control', 'experimental']

# Normalizar categorías
datos = tabla_cambios.copy()

datos['Grupo'] = (
    datos['Grupo'].astype(str).str.strip().str.lower()
)

datos['Variable'] = (
    datos['Variable'].astype(str).str.strip()
)

# Crear subgráficos
fig = make_subplots(
    rows=1,
    cols=2,
    subplot_titles=[
        variable for variable in variables
    ],
    horizontal_spacing=0.15
)

# Recorrer variables
for i, variable in enumerate(variables):

    columna = i + 1

    # Filtrar resultados de la variable
    datos_variable = datos[
        datos['Variable'] == variable
    ]

    # Recuperar resultados estadísticos
    resultado = tabla_efecto[
        tabla_efecto['Variable'] == variable
    ].iloc[0]

    prueba = resultado['Prueba aplicada']
    estadistico = resultado['Estadístico']
    pvalue = resultado['p-value']

    # Agregar diagramas de caja
    for grupo in grupos:

        valores = datos_variable.loc[
            datos_variable['Grupo'] == grupo,
            'Cambio'
        ].dropna()

        fig.add_trace(
            go.Box(
                y=valores,
                name=grupo.capitalize(),
                boxpoints='all',
                jitter=0.35,
                pointpos=0,
                boxmean=True,
                showlegend=(i == 0),
                hovertemplate=(
                    'Cambio: %{y}<br>'
                    'Grupo: ' + grupo.capitalize() +
                    '<extra></extra>'
                )
            ),
            row=1,
            col=columna
        )

    # Línea de referencia en cero
    fig.add_hline(
        y=0,
        line_dash='dash',
        line_color='gray',
        line_width=1.5,
        row=1,
        col=columna
    )

    # Anotación estadística
    fig.add_annotation(
        x=0.98,
        y=0.98,
        xref=f'x{columna} domain' if columna > 1 else 'x domain',
        yref=f'y{columna} domain' if columna > 1 else 'y domain',
        text=(
            f'{prueba}<br>'
            f'Estadístico = {estadistico:.3f}<br>'
            f'p-value = {pvalue:.4f}<br>'
            f'α = 0.05<br>'
            f'{"No se rechaza H₀" if pvalue > 0.05 else "Se rechaza H₀"}'
        ),
        showarrow=False,
        align='right',
        bgcolor='white',
        bordercolor='gray',
        borderwidth=1,
        borderpad=5,
        row=1,
        col=columna
    )

# Configuración general
fig.update_layout(
    title='Comparación de los cambios Pretest - Postest<br>Control vs Experimental',
    height=550,
    width=1200,
    template='plotly_white',
    boxmode='group'
)

fig.update_yaxes(
    title_text='Cambio (Postest - Pretest)'
)

fig.update_xaxes(
    title_text='Grupo'
)

fig.show()
```
